#### Question 1:

If $h^*$ is a minimizer of $\phi$, then

$$
\phi (h') \ge\phi(h^*) \quad \forall h' \in \mathcal{H}
$$

By definition of subgradient, $g \in \partial\phi(h^*)$ if

$$
\phi(h') \ge \phi(h^*) + \langle g, h' - h^* \rangle
$$

Let $g=0$, then

$$
\phi(h') \ge \phi(h^*) + \langle 0, h' - h^* \rangle. \\
\phi(h') \ge \phi(h^*)
$$

Hence

$$
0 \in \partial\phi(h^*)
$$

#### Question 2:

Let $h_1,h_2 \in \mathcal{H}$ be the minimizers of $\phi$ over $\mathcal{H}$.

By definition of $\alpha$-strong convexity, for any $g \in \partial\phi(h_1)$

$$
\phi(h_2) \ge \phi(h_1) + \langle g, h_2 - h_1 \rangle + \frac{\alpha}{2}\|h_2 - h_1\|^2
$$

From question 1, $0 \in \partial\phi(h_1)$, so let $g=0$

$$
\phi(h_2) \ge \phi(h_1) + \frac{\alpha}{2}\|h_2 - h_1\|^2
$$

Since both $h_1,h_2$ are minimizers

$$
\phi(h_1) = \phi(h_2)=\inf_{h\in\mathcal{H}} \phi(h)
$$

We have

$$
0 \ge  \frac{\alpha}{2}\|h_2 - h_1\|^2
$$

Since $\alpha>0$ and $\|h_2 - h_1\|^2$  cannot be negative, this implies $\|h_2 - h_1\|^2=0$, hence $h_1=h_2$. Thefore, the minimizer $h^*$ is unique

#### Question 3:

Since $h^*$ minimise $\phi$, $0 \in \partial\phi(h^*)$ and let $g=0$, by definition of $\alpha$-strong convexity

$$
\phi(\tilde{h}) \ge \phi(h^*) + \frac{\alpha}{2}\|\tilde{h} - h^*\|^2 \\
\phi(\tilde{h}) - \phi(h^*) \ge \frac{\alpha}{2}\|\tilde{h} - h^*\|^2
$$

Since $\tilde{h}$ minimise $\phi+\psi$, this implies

$$
 \phi(\tilde{h}) + \psi(\tilde{h}) \le \phi(h) + \psi(h) \quad \forall h \in \mathcal{H}
$$

The above inequality must also hold when $h=h^*$ is the minimizer of $\phi$, so we get

$$
\phi(\tilde{h}) + \psi(\tilde{h}) \le \phi(h^*) + \psi(h^*)
$$

Since $\psi$ is $L$-Lipschitz over $\mathcal{H}$, this implies

$$
| \psi(\tilde{h}) - \psi({h^*})| \le L\| \tilde{h} \ - h^* \|

$$

\frac{\alpha}{2}\|\tilde{h} - h^*\|^2 \le L\|\tilde{h} - h^*\|\\

#### Question 1:

Since $h \mapsto \ell(h, z)$ is $L$-Lipschitz for any $z \in Z$, this means

$$
|\ell(h, z) - \ell(h', z)| \le L\|h - h'\| \quad \forall h, h' \in \mathcal{H}
$$

Let

$$
f(h) = \frac{1}{n}(\ell(h, z_i) - \ell(h, z'_i))
$$

Then

$$
\begin{aligned}
|f(h) - f(h')| &= \frac{1}{n}|\ell(h, z_i) - \ell(h', z_i) - (\ell(h, z'_i) - \ell(h', z'_i))| \\
&\le \frac{1}{n}(|\ell(h, z_i) - \ell(h', z_i)| + |\ell(h, z'_i) - \ell(h', z'_i)|) \\
&\le \frac{1}{n}(L\|h - h'\| + L\|h - h'\|) = \frac{2L}{n}\|h - h'\|.
\end{aligned}
$$

Therefore, $f$ is Lipschitz with constant of

$$
\frac{2L}{n}
$$

#### Question 2:

\hat{R}_S(h) = \frac{1}{n}\sum_{i=1}^n \ell(h, z_i)

Define empirical risk on perturbed training data $S^i$

$$
\hat R_{S^i}(h) = \frac{1}{n}\left(\sum_{j \ne i} \ell(h, z_j) + \ell(h, z'_i)\right)

$$
\begin{aligned}
\hat R_S(\hat{h}_n^{(i)}) &\ge \hat R_S(\hat{h}_n) + \frac{\alpha}{2}\|\hat{h}_n^{(i)} - \hat{h}_n\|^2 \\
\hat R_{S^i}(\hat{h}_n) &\ge \hat R_{S^i}(\hat{h}_n^{(i)}) + \frac{\alpha}{2}\|\hat{h}_n^{(i)} - \hat{h}_n\|^2
\end{aligned}

Add above two equations together, we obtain

$$
\hat R_S(\hat{h}_n^{(i)}) - \hat R_S(\hat{h}_n) + \hat R_{S^i}(\hat{h}_n) - \hat R_{S^i}(\hat{h}_n^{(i)}) \ge \alpha\|\hat{h}_n^{(i)} - \hat{h}_n\|^2

#### Question 1:

Let

$$
\phi(h) = \ell(h, z) + \frac{\lambda}{2}\|h\|^2
$$

Since $\ell(h, z)$ is convex, for any $h, h' \in \mathcal{H}$ and for any subgradient $g \in \partial\ell(h, z)$,

$$
\ell(h', z) \ge \ell(h, z) + \langle g, h' - h \rangle
$$

Next, we do some manipulation to expand the quadratic term $\frac{1}{2}\|h\|^2$ in terms of any $h, h' \in \mathcal{H}$. Express the norm as inner product,

$$
\|h'\|^2 = \langle h', h' \rangle
$$

and by linearity of the inner product,

$$
\langle h', h' \rangle = \langle h + (h' - h), h + (h' - h) \rangle
$$

Expand the inner product,

$$
\begin{aligned}\|h'\|^2 &= \langle h, h \rangle + 2\langle h, h' - h \rangle + \langle h' - h, h' - h \rangle \\&= \|h\|^2 + 2\langle h, h' - h \rangle + \|h' - h\|^2.\end{aligned}
$$

Divide both side by 2 and multiply by $\lambda$ , we obtained

$$
\frac{\lambda}{2}\|h'\|^2 = \frac{\lambda}{2}\|h\|^2 + \langle \lambda h, h' - h \rangle + \frac{\lambda}{2}\|h' - h\|^2
$$

Combine the two inequalitiles, we obtained

$$
\begin{aligned}\phi(h',z) &= \ell(h',z) + \frac{\lambda}{2}\|h'\|^2 \\&\ge \ell(h,z) + \langle g, h' - h \rangle + \frac{\lambda}{2}\|h\|^2 + \langle \lambda h, h' - h \rangle + \frac{\lambda}{2}\|h' - h\|^2 \\&= \phi(h) + \langle g + \lambda h, h' - h \rangle + \frac{\lambda}{2}\|h' - h\|^2.\end{aligned}
$$

which is the same as the definition of strong convexity $(1)$ with constant $\alpha = \lambda$ and subgradient $g+\lambda h \in \partial \phi(h)$.

Hence,  $h \mapsto \ell(h, z) + \frac{\lambda}{2}\|h\|^2$ is $\lambda$-strongly convex

\le \frac{2(L+\lambda B)^2}{\lambda n} + \left(\frac{4(L+\lambda B)^2}{\lambda} + 2(1+\frac{\lambda}{2}B^2)\right)\sqrt{\frac{\log(4/\delta)}{2n}}
$$

Simplify my equation and let

$$

$$
C_1 = \frac{2}{n} + 4\sqrt{\frac{\log(4/\delta)}{2n}}, \quad C_2 = 2\sqrt{\frac{\log(4/\delta)}{2n}}
$$

Differentiate $f$ with respect to $\lambda$

$$
f(\lambda) = C_1\left(\frac{L^2}{\lambda} + 2LB + B^2\lambda\right) + C_2\left(1 + \frac{\lambda}{2}B^2\right)
$$

$$
f'(\lambda) = -C_1\frac{L^2}{\lambda^2} + B^2\left(C_1 + \frac{C_2}{2}\right)
$$

Set $f'(\lambda) = 0$ to find the minimiser

$$
-C_1\frac{L^2}{\lambda^2} + B^2\left(C_1 + \frac{C_2}{2}\right) = 0 \\
\lambda^2 = \frac{C_1 L^2}{B^2(C_1+\frac{C_2}{2})}
$$

Hence the optimal $\lambda$ is

$$
\lambda^* = \frac{L}{B}\sqrt{\frac{C_1}{C_1+\frac{C_2}{2}}} \\
\lambda^* \approx \frac{L}{B}
$$

Now with

L+\lambda B = 2L, \quad \lambda = L/B, \quad \lambda B^2 = LB,

$$
\frac{2(L+\lambda B)^2}{\lambda n} = \frac{2(2L)^2}{(L/B)n} = \frac{8LB}{n}

$$

We get the optimal bound

R(\hat{h}_{n,\lambda}) - \inf_h\left\{R(h) + \frac{\lambda}{2}\|h\|^2\right\} \le \frac{8LB}{n} + (17LB+2)\sqrt{\frac{\log(4/\delta)}{2n}}

```python
class BaseNetwork(nn.Module):

    def __init__(self, act_fn, input_size, output_size, hidden_sizes):
        """
        Inputs:
            act_fn - Object of the activation function that should be used as non-linearity in the network.
            input_size - Size of the input images in pixels
            output_size - Number of classes we want to predict
            hidden_sizes - A list of integers specifying the hidden layer sizes in the NN
        """
        super().__init__()

        # Initialise layers
        layers = []

        # Add input layer and activation function to layers list
        layers += [nn.Linear(input_size, hidden_sizes[0]), act_fn]

        # Add hidden layers to layers list and activation function after each layer
        for i in range(1, len(hidden_sizes)):
            layers += [nn.Linear(hidden_sizes[i - 1], hidden_sizes[i]), act_fn]

        # Add output layer to layers list (no activation function after output layer)
        layers += [nn.Linear(hidden_sizes[-1], output_size)]

        # Combine into Sequential model
        self.layers = nn.Sequential(*layers)

    def forward(self, x):

        # Flatten the input
        x = x.view(x.size(0), -1)

        # Pass the input through the layers
        out = self.layers(x)

        return out
```

```python
class Optimizer:

    def __init__(self, mynet, X, y, learning_rate, tolerance=1e-6):
        '''
        Initializes the Optimizer.

        Parameters
        ----------
        mynet : nn.Module
            The neural network model to be optimized.
        X : 2D torch Tensor
            Input data with dimensions (n_samples, n_features).
        y : 1D torch Tensor
            Output/target data with dimension (n_samples,).
        learning_rate : float
            Learning rate or regularization parameter.
        tolerance : float, optional
            Tolerance for convergence criteria.
        '''

        # Storing initial data
        self.X = X
        self.y = y

        # Extracting dimensions
        self.n_sample, self.n_feature = X.shape

        self.net = mynet

        self.learning_rate = learning_rate

        self.tolerance = tolerance

        # Algorithm history
        self.obj_history = [self.myloss(self.X, self.y).item()]

    def myloss(self, X, y):

        # Mean squared error
        loss = torch.mean((self.net(X)- y) ** 2)
        return loss

    def plot_loss(self):

        if len(self.obj_history) == 1:
            raise ValueError('No history to plot.')

        fig, ax = plt.subplots(figsize=(8, 4))
        ax.plot(self.obj_history, color='k')
        ax.set_ylabel('Objective Function Value')
        ax.set_xlabel('Iteration')
        ax.grid()
        fig.tight_layout()
        plt.show()
```


#### Question 4

```python
class GradientDescentOptimizer(Optimizer):

    def __init__(self, mynet, X, y, learning_rate, tolerance=1e-6, max_iters=1000):
        super().__init__(mynet, X, y, learning_rate, tolerance)
        self.max_iters = max_iters

    def run(self):

        n_iter = 0
        while n_iter < self.max_iters:

            # Compute MSEloss (forward pass)
            loss = self.myloss(self.X, self.y)

            # Reset gradients to zero before backward pass
            self.net.zero_grad()

            # Compute gradients using PyTorch autograd
            loss.backward()

            # Update each parameter manually
            with torch.no_grad():
                for p in self.net.parameters():
                    p -= self.learning_rate * p.grad

            # Track loss value
            new_loss = self.myloss(self.X, self.y).item()
            self.obj_history.append(new_loss)

            n_iter += 1

            # Convergence check (small loss change)
            if abs(self.obj_history[-2] - self.obj_history[-1]) < self.tolerance:
                print(f"Converged in {n_iter+1} iterations")
                break

        if n_iter == self.max_iters:
            print(f"Reached maximum iterations ({self.max_iters}).")

```

```python
class AcceleratedGradient(Optimizer):

    def __init__(self, mynet, X, y, learning_rate, tolerance=1e-6, max_iters=1000):
        super().__init__(mynet, X, y, learning_rate, tolerance)
        self.max_iters = max_iters

    def run(self):

        # Initialize momentum variables
        t_k = torch.tensor(0.0)
        x_prev = [p.detach().clone() for p in self.net.parameters()]
        x_k = [p.detach().clone() for p in self.net.parameters()]

        n_iter = 0
        while n_iter < self.max_iters:

            # Compute new momentum term
            t_next = 0.5 * (1 + torch.sqrt(1 + 4 * t_k**2))

            # Compute extrapolated term
            coeff = (t_k - 1) / t_next
            x_bar = [xk + coeff * (xk - x_prev) for xk, x_prev in zip(x_k, x_prev)]

            # Load extrapolated parameters into the network
            with torch.no_grad():
                for p, xb in zip(self.net.parameters(), x_bar):
                    p.copy_(xb)

            # Compute loss and gradient
            loss = self.myloss(self.X, self.y) # MSELoss
            self.net.zero_grad()               # Zero before backward pass
            loss.backward()                    # PyTorch autograd

            # Update each parameter manually
            with torch.no_grad():
                for p in self.net.parameters():
                    p -= self.learning_rate * p.grad

            # Update stored parameters and momentum
            x_prev = x_k
            x_k = [p.detach().clone() for p in self.net.parameters()]
            t_k = t_next
            n_iter += 1

            # Track new loss value
            new_loss = self.myloss(self.X, self.y).item()
            self.obj_history.append(new_loss)

            # Convergence check (small loss change)
            if abs(self.obj_history[-2] - self.obj_history[-1]) < self.tolerance:
                print(f"Converged in {n_iter} iterations")
                break

        if n_iter == self.max_iters:
            print(f"Reached maximum iterations ({self.max_iters}).")
```

```python
# Define model with SELU
mynet = BaseNetwork(
    act_fn=nn.SELU(),
    input_size=X_train.shape[1],
    output_size=y_train.shape[1],
    hidden_sizes=[32, 16, 8]
)

# Run accelerated gradient
ag = AcceleratedGradient(
    mynet=mynet,
    X=X_train,
    y=y_train,
    learning_rate=1e-3,
    tolerance=1e-6,
    max_iters=1000
)

ag.run()

# Compute validation loss
val_loss_ag = ag.myloss(X_val, y_val).item()
print(f"Validation loss (Accelerated Gradient): {val_loss_ag:.6f}")

ag.plot_loss()
```

Using the same network architecture, SELU activation function, learning rate, max. iterations,
- With standard gradient descent, the validation loss after 1000 iterations was approximately 0.96.
- With accelerated gradient method, the validation loss after 1000 iterations was significantly lower, around 0.69.

This means that, for the same computational cost, accelerated method achieves a better minimisation of the loss and better generalisation performance on the validation set
